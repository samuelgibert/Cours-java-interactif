# La structure de base d'un programme en JAVA

La structure générale d'un programme est la suivante :

````java
/** commentaire d'entête permettant d'expliquer 
1. Le rôle du programme
2. L'auteur du programme
3. La date de création
4. La date de mise à jour
Etc..
*/

//Zone d'importation des classes utilisées dans le fichier
import … ;

public class Appli {  //ici Appli est le nom de la classe, mais il peut être remplacé par un autre nom, à condition que le nom du fichier soit le même que celui de la classe

    public static void main(String[] args) {   //méthode main, point d'entrée du programme : ne pas modifier cette ligne.

 		/* zone de déclaration des variables locales */
		Declaration1;
		Declaration2;

 		/* CORPS DU PROGRAMME */
		Instruction1;
		Instruction2;
}   
} /* FIN */
````

Un programme JAVA s'écrit dans un fichier texte ayant comme extension « **.java** ». Le nom du fichier doit porter le nom de la classe. Dans l'exemple ci-dessus, la classe s'appelle **Appli**, donc ce programme doit obligatoirement se trouver dans un fichier nommé « **Appli.java** »

## Les commentaires

Les commentaires permettent une relecture de programme pour le programmeur lui-même, mais aussi pour ceux qui vont relire le programme (travail en équipe). Un commentaire d'entête est nécessaire pour indiquer le nom du programmeur, la date, le nom du fichier, descriptif du programme. Les commentaires sont du texte simple et ne sont donc pas traités par le compilateur.

2 types de commentaires :

- // indiquent au compilateur que le reste de la ligne est un commentaire.

- /* ….  */ indiquent au compilateur qu'à partir de /* le texte est un commentaire jusqu'au symbole */ permet des commentaires sur plusieurs lignes.

La plupart des IDE ajoute automatiquement un commentaire d'entête, comme son nom l'indique en début de fichier.

````java
/** 
 * File:   Appli.java
 * Author: sgibert
 *
 * Created on 14 septembre 2018, 20:19
 * 
 * Description du programme : explique ici ce que fait ton programme 
 */
````

## Le mécanisme du cast

Ce mécanisme permet au langage JAVA de gérer les problèmes de compatibilité de type. Pour illustrer le principe imaginez que vous cherchier à transvaser un verre d'eau dans un autre verre sans perdre de l'eau

![image4](image3.png)



Comme illistrer avec l'image des verres, transverser un verre d'eau est **possible** dans les 2 cas suivants:

- Les 2 verres sont de même contenance. Donc si 2 variables sont 2 même type. Par exemple int avec int, short avec short, etc ...

- transvaser un petit verre d'eau dans un autre verre plus grand. Par exemple **short** -> **int**, ou **int** -> **long**.



````java
//TESTER CE CODE POUR COMPRENDRE LE FONCTIONNEMENT DES AFFECTATIONS ENTRE TYPES
short a = 500;
int b = 1000;
long c = 10000000000L;
int d;
d=b;  //affectation d'une variable de type int à une variable de type int, pas de problème car c'est le même type
b=a;  //affectation d'une variable de type short à une variable de type int, pas de problème car int est plus grand que short
c=b;  //affectation d'une variable de type int à une variable de type long, pas de problème car long est plus grand que int
System.out.println(b);  
````

Dans les autres cas c'est **impossible** de transvaser un "grand" vers d'eau vers un "petit" sans perdre de l'eau. Le lanagage JAVA n'autorise pas ce type d'opération.

C'est à ce moment là qu'on utilise le principe du cast pour forcer ce qui est impossible. Mais attention cela peut entrainer des pertes de données. Avec le cast on "transforme" une variable d'un certain type vers un autre.

La syntaxe du cast est la suivante:

```java
         variable = (typeVariable) expression;    //expression peut désigner une autre variable, une valeur, ou bien encore une expression arithmétique
``` 

Prenons l'exemple suivant en apportant la correction nécessaire en fonction des commentaires

````java
//CORRIGER LES ERREURS D'AFFECTATION ENTRE TYPES
byte b=128;     // la valeur 128 est considérée comme un int et de plus dépassant les limites d'un byte, 
                // Cette valeur est affectée à une variable de type byte. Erreur ==>  Il faut donc faire un cast pour forcer 128 a devenir un byte   
int a = 10;
b=a;            // Erreur : impossible, a est une int donc de capacité > byte. il faut donc faire un cast pour forcer "a" devenir un byte, 
                // meme si la valeur de "a" est inférieure à 127.
````

Il est possible de transformer une valeur d'un type dans un autre type grâce au mécanisme du **cast**.

````java
//TESTER CE CODE POUR COMPRENDRE LE FONCTIONNEMENT DES AFFECTATIONS ENTRE TYPES
byte b=127;
int a=1500;        
b=(byte)a;    // a est convertie en byte et affectée à b, mais cela entrainte une perte à la conversion puisque cette valeur dépasse la valeur max d'un byte.
			 // donc b ne vaut pas 1500, mais une autre valeur qui est le résultat de la conversion de 1500 en byte. En effet, le type byte ne peut contenir que des valeurs entre -128 et 127.
        
System.out.println("" + b);	 // affiche -36
````

L'ordre de capacité est le suivant : 		**byte < short < int < long < float < double.**

Voici un autre exemple :

````java
//TESTER CE CODE POUR COMPRENDRE LE FONCTIONNEMENT DES AFFECTATIONS ENTRE TYPES
int i = 2;
float fa = 5.56f;

i = (int) fa;    	// fa est convertie en int, i vaut 5. En effet, le type int ne peut contenir que des valeurs entières, donc la partie décimale est perdue.
System.out.println("" + i);	 // affiche 5
fa = (float) i;	    // i est converti en float, fa vaut 5.0. Ici le cast (float) est inutile, car i est déjà un int, et l'affectation d'un int à un float se fait automatiquement.
System.out.println("" + fa);	 // affiche 5.0
````

- Cas particulier pour l'opérateur +, -, * , / , % : Pour les types entiers (byte, short, int et long), Java met en œuvre un mécanisme de conversion implicite vers le type int appelé promotion entière. Cela oblige à effectuer des cast pour les types byte et short.

````java
byte a=5,b=127,c=0;  
c = a + b;   // le résultat de l'opération a + b est converti automatiquement (promotion vers int) en int puis affecté à un byte : erreur
````

Dans cet exemple (a) et (b) sont de type « byte ». L'opération d'addition ne donne pas un **byte,** mais un **int** du fait de la conversion implicite. Cela provoque donc une erreur de compilation. Pour corriger ce problème, il faut effectuer un **cast** au niveau de l'opération. On écrira :

````java
c = (byte)(a + b);              // Attention : affiche ??? 
System.out.println("c="+c);     //affiche -124, car le résultat de l'opération + est converti automatiquement en int, puis converti en byte, ce qui entraîne une perte de données. 
                                // En effet, le type byte ne peut contenir que des valeurs entre -128 et 127.
````

- Cas particulier de la division : Il ne faut jamais faire de division par 0. À l'exécution le programme génère une Exception du type : « Exception in thread "main" java.lang.ArithmeticException: / by zero »

Exemple :

````java
int a=10,b=0;
c=a/b;   //ici b vaut 0, donc division par 0

int a=10,b=2;
c=a/b;   //ici b vaut 2, aucun problème
````

- L'opérateur / sert à la fois pour diviser des entiers (byte, short, int, long)  et des réels (float et double). En fonctions du type, le comportement est différent. Cela pose parfois des problèmes de résultat.

- Exemple:

````java
//TESTER CE CODE POUR COMPRENDRE LE FONCTIONNEMENT DE LA DIVISION ENTRE ENTIER ET REEL
int a=5,b=2;
float c;   	// c est un réel
c = a / b;	// comme a et b sont des entiers, le résultat  de la division est un entier
 		  	// donc c ne vaut pas 2.5, mais 2.0
System.out.println("c="+c);	//affiche 2.0
````

Pour pallier à ce problème, il existe 2 solutions :

- Mettre b en float, ce qui revient à diviser un entier (a) par un réel (b), dans ce cas (c) vaut 2.5

````java
int a=5;
float c,b=2;
c=a/b;  // division d'un entier par un réel. Le résultat est un réel
System.out.println("c="+c);	//affiche 2.5
````

- Faire ce qu'on appelle un **CAST** ou conversion de type. On laisse (b) en entier, mais au moment de la division (b) est convertie en float:

Ce type d'opérateur dans les tests des conditions (if, else if) ou des boucles (for, while, do while). Le résultat d'une opération logique est une valeur booléenne (true ou flase)

| Opérateur | Signification |

Exemple :

````java
int i=4,j=3 ;
boolean v ;
v=(i<=j) ;              // v vaut ?
System.out.println("v="+v);
v=(i==4 && i > 2);      // v vaut ? 
System.out.println("v="+v);
````

## Les opérateurs d'affectation et d'incrémentation

| Opérateur | Signification |
|---|---|
| = | Affectation |
| += | Ajout de l'opérande de droite |
| -= | Suppression de l'opérande de droite |
| *= | Multiplication de l'opérande de droite |
| /= | Division de l'opérande de droite |
| %= | Modulo de l'opérande de droite |
| ? : | Affectation conditionnelle |
| ++variable | Préincrémentation (avant) |
| variable++ | Postincrémentation (après) |
| --variable | Prédécrémentation |
| variable-- | Postdécrémentation |

Quelques exemples simples:

````java
int i=3;
i++ ;       // i=i+1 ;
i-- ;       // i=i-1 ; 
i+=1 ;      //i=i+1 ;
i*=5;       //i=i*5 ;
````

Attention, certains opérateurs sont dits à effet de bord, et s'ils sont mal utilisé (ou mal maitrisé) peuvent donner des résultats inattendus. Exemple

````java
//
int a=5,b=1,c=0;
c=++b + a++;   //c vaut ?
System.out.println("c="+c);  //affiche 7, car b est incrémenté avant l'addition, donc b vaut 2, et a est incrémenté après l'addition, donc a vaut 5 lors de l'addition. Donc c = 2 + 5 = 7.

int a=5,b=1,c=0;
c=b++ + a++;   //affiche 6, car b est incrémenté après l'addition, donc b vaut 1 lors de l'addition, et a est incrémenté après l'addition, donc a vaut 5 lors de l'addition. Donc c = 1 + 5 = 6.
System.out.println("c="+c);
````

Conseils : Programmez au début en utilisant une façon simple d'incrémenter et d'affecter.

L'affectation conditionnelle :

*Syntaxe* : **<opérande> = <condition> ?<expression1> :<expression2> ;**

Si condition vaut true alors on affecte à opérande expression1, si false expresson2.

Exemple :

````java
int i=3,j=4,k ;
k= i>j ?10 :-10 ;
System.out.println("k="+k);  //affiche -10, car i>j est faux, donc k vaut -10
````

## PREMIÈRE NOTION DE PACKAGE

Tout ce qui est nécessaire pour la programmation de base en java est inclus dans un package automatiquement importé.

````java
import java.lang.*;    //cette ligne n'est pas obligatoire, car automatiquement ajouté. 
````

Un package est un ensemble de classes. En fait, c'est un ensemble de dossiers et de sous-dossiers contenant une ou plusieurs classes. Charger un package nous permet d'utiliser les classes qu'il contient. ## La classe Scanner  pour la lecture

Une lecture de donnée se fait sur l'entrée standard en Java qui correspond à System.in

Mais pour faire cette lecture, il nous faut un objet de type Scanner. C'est une classe définie dans le package java.util

Exemple :

````java
import java.util.Scanner; //ici on importe juste la classe Scanner
import java.util.*; //ici on importe toutes les classes de java.util

Scanner sc = new Scanner (System.in);   //sc est un objet de type Scanner. Il va lire sur l'entrée standard.  
````

Maintenant que nous avons notre objet, il est possible d'effectuer des lectures au clavier grâce à différentes méthodes.

Par exemple :

````java
System.out.println("Saisir une phrase");
String str=sc.nextLine();
System.out.println(""+str);

System.out.println("Saisir un nombre");
int a= sc.nextInt();
System.out.println("votre nombre "  +a);
````

Attention toutefois, dans le dernier exemple la méthode nextInt() peut provoquer une erreur, notamment si l'entrée au clavier ne peut être convertie en entier. Nous verrons plus tard comment gérer ce type de problème au travers du mécanisme d'exceptions.
